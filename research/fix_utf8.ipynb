```python
from pathlib import Path

data_dir = Path('../data/')

for file in data_dir.iterdir():
    if file.name.startswith('elect'):
        encoding_read = 'utf-8'
        encoding_write = 'utf-8-sig'
        print(f"File: {file.name} (encoding: {encoding_read})")
        df = pd.read_csv(file, encoding=encoding_read)
        print(df.head(3))
        df.to_csv(file.with_suffix('.utf8.csv'), index=False, encoding=encoding_write)
```
